```python
import numpy as np
import matplotlib.pyplot as plt

from DMRates import *

import wimprates
import numericalunits as nu
```

```python
""" Set MChi to be 100 GeV """
MChi_GeV = 1000
""" Set cross section with nucleon """
sigma0 = 1E-40 * CM2

""" Get event rate function """
EventRate = XeSpinIndependentRate(MChi_GeV, sigma0, delta_keV=100)
""" x axis """
X = np.linspace(0, 200, 10000)
""" Corresponding recoil energy """
ER = X * KEV
""" Total event rate in events/ton/keV/year """
Y = [EventRate(Er) * DAY * KEV * KG for Er in ER]

plt.plot(X, Y)

plt.yscale('log')
plt.xlabel('Recoil Energy [keV]')
plt.ylabel('Rate [events / (keV kg day)]')
plt.xlim(0, 200)
plt.ylim(1e-12, 1)
plt.grid()
plt.legend()
plt.show()
```

```python
es = np.linspace(1e-10, 200, 10000)

halo_model = wimprates.StandardHaloModel(v_0=220, v_lab=232)

dr_de_100 = wimprates.rate_wimp_std(
    es,
    mw=1000,       # mass of wimp (GeV)
    sigma_nucleon=1e-40, # cross section (cm^2)
    material='Xe', # target material
    halo_model=halo_model,
    detection_mechanism='inelastic_nr',
    delta=100
) / 1000 / (nu.year/nu.day)


plt.plot(X, Y, label='DMRates')
plt.plot(es, dr_de_100, label='delta = 100 keV', linestyle='--')
plt.yscale('log')
plt.xlabel('Recoil Energy [keV]')
plt.ylabel('Rate [events / (keV kg day)]')
plt.xlim(0, 200)
plt.ylim(1e-12, 1)
plt.legend()
plt.grid()
plt.show()
```
